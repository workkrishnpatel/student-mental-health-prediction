```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib as mpl


# DARK THEME SETTINGS

plt.style.use("dark_background")

sns.set_theme(style="dark")


# FORCE COLORS

mpl.rcParams['figure.facecolor'] = '#0d1117'

mpl.rcParams['axes.facecolor'] = '#0d1117'

mpl.rcParams['savefig.facecolor'] = '#0d1117'

mpl.rcParams['axes.edgecolor'] = 'white'

mpl.rcParams['axes.labelcolor'] = 'white'

mpl.rcParams['xtick.color'] = 'white'

mpl.rcParams['ytick.color'] = 'white'

mpl.rcParams['text.color'] = 'white'

mpl.rcParams['grid.color'] = 'gray'
```

```python
df = pd.read_csv(R"C:\Users\workk\OneDrive\Desktop\Mental Health Risk Prediction\Student Depression Dataset.csv")
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x='Depression',
    data=df
)

plt.title("Depression Distribution", fontsize=18)

plt.xlabel("Depression")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='Gender',
    hue='Depression',
    data=df
)

plt.title("Gender vs Depression", fontsize=18)

plt.xlabel("Gender")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.gca().set_facecolor('#0d1117')

sns.histplot(
    df['Academic Pressure'],
    bins=10,
    kde=True,
    color='cyan'
)

plt.title("Academic Pressure Distribution", fontsize=18)

plt.xlabel("Academic Pressure")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.gca().set_facecolor('#0d1117')

sns.boxplot(
    x='Depression',
    y='Academic Pressure',
    data=df,
    palette='viridis'
)

plt.title("Depression vs Academic Pressure", fontsize=18)

plt.xlabel("Depression")

plt.ylabel("Academic Pressure")

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.gca().set_facecolor('#0d1117')

sns.countplot(
    x='Sleep Duration',
    hue='Depression',
    data=df,
    palette='magma'
)

plt.xticks(rotation=45)

plt.title("Sleep Duration vs Depression", fontsize=18)

plt.xlabel("Sleep Duration")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.gca().set_facecolor('#0d1117')

sns.countplot(
    x='Dietary Habits',
    hue='Depression',
    data=df,
    palette='plasma'
)

plt.title("Dietary Habits vs Depression", fontsize=18)

plt.xlabel("Dietary Habits")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.gca().set_facecolor('#0d1117')

sns.countplot(
    x='Have you ever had suicidal thoughts ?',
    hue='Depression',
    data=df,
    palette='coolwarm'
)

plt.title("Suicidal Thoughts vs Depression", fontsize=18)

plt.xlabel("Suicidal Thoughts")

plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.gca().set_facecolor('#0d1117')

sns.boxplot(
    x='Depression',
    y='Financial Stress',
    data=df,
    palette='magma'
)

plt.title("Financial Stress vs Depression", fontsize=18)

plt.xlabel("Depression")

plt.ylabel("Financial Stress")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.gca().set_facecolor('#0d1117')

sns.boxplot(
    x='Depression',
    y='Work/Study Hours',
    data=df,
    palette='viridis'
)

plt.title("Work/Study Hours vs Depression", fontsize=18)

plt.xlabel("Depression")

plt.ylabel("Work/Study Hours")

plt.show()
```

```python
temp_df = df.copy()
```

```python
from sklearn.preprocessing import LabelEncoder

for col in temp_df.columns:

    if temp_df[col].dtype == 'object':

        le = LabelEncoder()

        temp_df[col] = le.fit_transform(temp_df[col])
```

```python
plt.figure(figsize=(14,10))

plt.gca().set_facecolor('#0d1117')

sns.heatmap(
    temp_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Feature Correlation Heatmap", fontsize=20)

plt.show()
```
